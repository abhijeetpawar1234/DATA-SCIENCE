# data collection 

```python
with open("finaldata.txt", encoding = 'utf-8') as f:
    data = f.read()
    
```

```python
data
```

```python
chunks = data.split("\n\n")
chunks = [i for i in chunks if len(i)>3]
```

```python
def parse_chunk(chunk):

        chunk = chunk.strip()
        sep_chunk = chunk.split('\n')
        username = sep_chunk[0]
        no_of_posts = int(sep_chunk[1].split(" post")[0].replace(",",""))
        followers = float(sep_chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M",""))
        if "K" in sep_chunk[2]:
            followers =int(followers * 1000)
        elif "M" in sep_chunk[2]:
            followers =int(followers * 1000000)
        else:
            followers =int(followers)
            
        
    
        following = float(sep_chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M",""))
        if "K" in sep_chunk[3]:
            following =int(following * 1000)
        elif "M" in sep_chunk[3]:
            following =int(following * 1000000)
        else:
            following =int(following)
            
        name = sep_chunk[4]
        if(len(sep_chunk)>5):
            type_of_page = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_page = "Unknown"
            bio = ""
            
    
        # print(username, no_of_posts, followers, following, name, type_of_page, bio, sep='\n')
        return {"username":username,"no_of_posts":no_of_posts,"followers":followers,"following":following,"name":name,"type_of_page":type_of_page ,"bio":bio }
        


# print(parse_chunk(chunks[0]))

    
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
print(all_chunks)
```

```python
import json
s = json.dumps(all_chunks,indent=4)
print(s)

with open("data.json",'w') as f:
    f.write(s)
```

```python
def max_followers(all_chunks):
    max = 0
    for chunk in all_chunks:
        if max < chunk['followers']:
            max = chunk['followers']
            max_followers = chunk
    print(max_followers['username'])

def max_following(all_chunks):
    maxfollo = 0
    for chunk in all_chunks:
        if maxfollo < chunk['following']:
            maxfollo = chunk['following']
            max_following_id = chunk
    print(max_following_id['username'])

category = set()
def no_of_category(all_chunks):
    for chunk in all_chunks:
        category.add(chunk['type_of_page'])
    print(category,f"there are :{len(category)} category")
    
print("maximum followers:")
max_followers(all_chunks)
print("maximum following:")
max_following(all_chunks)

print("number of category:")
no_of_category(all_chunks)

```
