Slicing returns a view, not a copy! Changes affect the original array.**
This might seem counterintuitive since Python lists create copies when sliced. But in NumPy, slicing returns a view of the original array. Both the sliced array and the original array share the same data in memory, so changes in the slice affect the original array. with copy() to make copy of existing array

```python
flat
b = flat[3:6] #b = flat[3:6].copy()

b[0] = 44444
```

```python
# Fancy Indexing (Select Multiple Elements)
arr = np.array([10,20,30,40])
ind = [0,1,2]
print(arr[ind])
```

Boolean Masking (Filter Data)

```python
arr2 = np.array([10, 20, 30, 40, 50])
mask = arr2 > 25  # Condition: values greater than 25
print(arr2[mask])  # [30 40 50]
```

exercise questions:
Create a 3×3 array filled with random numbers and print its shape.
Convert an array of floats [1.1, 2.2, 3.3] into integers.
Use fancy indexing to extract even numbers from [1, 2, 3, 4, 5, 6].
Reshape a 1D array of size 9 into a 3×3 matrix.
Use boolean masking to filter numbers greater than 50 in an array.

```python
import numpy as np

# 1. Create a 3×3 array filled with random numbers and print its shape
arr_random = np.random.rand(3, 3)
print("Random Array:\n", arr_random)
print("Shape:", arr_random.shape)

# 2. Convert an array of floats [1.1, 2.2, 3.3] into integers
float_arr = np.array([1.1, 2.2, 3.3])
int_arr = float_arr.astype(int)
print("\nInteger Array:", int_arr)

# 3. Use fancy indexing to extract even numbers from [1, 2, 3, 4, 5, 6]
arr = np.array([1, 2, 3, 4, 5, 6])
even_numbers = arr[[1, 3, 5]]  # indices of even values
print("\nEven Numbers (Fancy Indexing):", even_numbers)

# 4. Reshape a 1D array of size 9 into a 3×3 matrix
one_d = np.arange(1, 10)
matrix = one_d.reshape(3, 3)
print("\n3x3 Matrix:\n", matrix)

# 5. Use boolean masking to filter numbers greater than 50 in an array
numbers = np.array([10, 55, 23, 78, 49, 90, 12])
filtered = numbers[numbers > 50]
print("\nNumbers > 50:", filtered)
```
