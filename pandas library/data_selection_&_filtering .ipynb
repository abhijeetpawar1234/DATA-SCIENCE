# Select Specific Rows and Columns

```python
df.loc[8,'IMDb'] # Value at row 8, column 'IMDb'
```

```python
df.iloc[7,5] # Value at row 7, column at index 5
```

```python
df.loc[0:2,["Actor","IMDb"]] # Rows 0 to 2, selected columns
```

```python
df.iloc[0:2,0:2] # Rows and cols by index position
```

# Fast Access: .at and .iat

```python
df.at[0,"Actor"] # Fast label-based access
```

```python
df.iat[0,0] # Fast position-based access
```

# Filtering with Conditions

```python
df[df["IMDb"]>6]
```

```python
df[(df["IMDb"] > 6) | (df["Year"] > 2019)]
```

# Querying with .query()

```python
df.query("Year >= 2019 and IMDb > 8")
```

```python
df.query("Actor == 'Kartik Aaryan' ")
```
