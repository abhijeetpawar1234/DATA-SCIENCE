```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
pd.merge(employees,departments, on="DeptID") # inner join 
```

```python
pd.merge(employees,departments, on="DeptID",how="left") # left join 
```

```python
pd.merge(employees,departments, on="DeptID",how="right") # right join
```

```python
pd.merge(employees,departments, on="DeptID",how="outer") # outerjoin Includes all data, fills missing with NaN.
```
